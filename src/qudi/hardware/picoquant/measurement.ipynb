# Looping the spacing

```python
import ppg512  # or adjust import to your path


PPG = ppg512.PPG512(qudi_main_weakref=None, name='ppg512')

# Set config options explicitly
PPG._port = 'COM3'
PPG._vccrf = 12000
PPG._vref = 400

PPG.on_activate()

#spacing = np.array([5,10,15,20,25,30,40,50,60,70,80,90])*5
spacing = np.linspace(5, 45,41)*5
#if spacing greater than 40 use 10ns spacing steps
try:
    for s in spacing:
        voltages = PPG.wg.create_gaussian_train(width=1, posi0=1, n=2, spacing=s)
        print(s/5)
        time.sleep(1)  # Allow some time for the waveform to be created
        ans = PPG.write_waveform(voltages=voltages)
        print(f"Device response: {ans}")         
finally:    
    PPG.on_deactivate()
```

# Constant Waveform

```python
import ppg512  # or adjust import to your path
import numpy as np
import time

# Initialize the device
PPG = ppg512.PPG512(qudi_main_weakref=None, name='ppg512')
PPG._port = 'COM3'
PPG._vccrf = 12000
PPG._vref = 400
PPG.on_activate()

try:
    voltages = PPG.wg.create_gaussian_train(width=1, posi0=1, n=2, spacing=30)
    ans = PPG.write_waveform(voltages=voltages)
    print("Waveform sent with spacing 30")
    while True:
        time.sleep(1)
except KeyboardInterrupt:
    print("Interrupted by user. Deactivating device...")
    PPG.on_deactivate()
```
