```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.optimize import curve_fit
```

```python
calib = np.genfromtxt(r"Z:\Oliver\measurements\238\calibration-spectrometer.txt")
```

```python
def func(xx, a, b):
    return a*xx+b

wl_laser = calib[:,0]
wl_spec = calib[:,1]
XX = np.linspace(min(wl_spec), max(wl_spec),100)

popt, pcov = curve_fit(func, wl_spec, wl_laser)

plt.plot(wl_spec, wl_laser, 'o', label='data')
plt.plot(XX, func(XX,*popt), label = '%.2f*x+%.2f'%(popt[0], popt[1]))
plt.legend()
plt.xlabel('lambda spec')
plt.ylabel('lambda laser')
```

```python
class talk_to_devices():
    def __init__(self):
        powermeter.set_activity_state('Power', True)
        return
    
    def _start_power_channel(self):
        powermeter.set_activity_state('Power', True)
        return

    def _stop_power_channel(self):
        powermeter.set_activity_state('Power', False)
        return

    def get_power(self):
        # in W
        time.sleep(0.5)
        power = powermeter.get_process_value('Power')
        time.sleep(0.5)
        power = power*1e6 # convert to uW
        return power
    
    def rotate_filterwheel(self, pos, limits=(0,360)):
        min_pos = limits[0]
        max_pos = limits[1]
        pos = min_pos if pos <= min_pos else pos
        pos = max_pos if pos >= max_pos else pos
        motordriver_pi3_remote.current_motor = 1
        motordriver_pi3_remote.motor_position = pos
        time.sleep(1.5)
        return pos
    
    def adjust_power(self, target_power, step_size=1, epsilon=0.5, limits_filterwheel=(-100,460)):
        
        actual_power = self.get_power()
        actual_posi = motordriver_pi3_remote.motor_position
        target_posi = actual_posi
        minval = actual_power
        minpos = actual_posi
        maxval = actual_power
        maxpos = actual_posi
        maxtries = int(360/abs(step_size))+2
        tries = 0
        if actual_power >= target_power+epsilon:
            # power too big, need to decrease power, need to increase angle
            print('decreasing power')
            step = abs(step_size)
            while actual_power >= target_power+epsilon:
                tries += 1
                target_posi = target_posi +  step
                print('moving to posi %.1f°'%target_posi)
                actual_posi = self.rotate_filterwheel(target_posi, limits=limits_filterwheel)
                actual_power = self.get_power()
                if actual_power <= minval:
                    minval = actual_power
                    minpos = actual_posi
                if tries >= maxtries:
                    # go to min posi if one full rotation
                    actual_posi = self.rotate_filterwheel(minpos, limits=limits_filterwheel)
                    time.sleep(2)
                    actual_power = self.get_power()
                    print('Could not reached desired value. Went to minimum power %fuW'%actual_power)
                    return actual_power
                # if actual_posi >= limits_filterwheel[1]-step_size:
                #    print('Reached lower limit of filterwheel position, power is %fuW'%actual_power)
                #    return actual_power
            print('reached target: %fuW.'%actual_power)
            return actual_power
        elif actual_power <= target_power-epsilon:
            # power too small, need to increase power, need to decrease angle
            print('increasing power')
            step = -abs(step_size)
            while actual_power <= target_power-epsilon:
                tries += 1
                target_posi = target_posi +  step
                print('moving to posi %.1f°'%target_posi)
                actual_posi = self.rotate_filterwheel(target_posi, limits=limits_filterwheel)
                actual_power = self.get_power()
                if actual_power >= maxval:
                    maxval = actual_power
                    maxpos = actual_posi
                if tries >= maxtries:
                    # go to max posi if one full rotation
                    actual_posi = self.rotate_filterwheel(maxpos, limits=limits_filterwheel)
                    time.sleep(2)
                    actual_power = self.get_power()
                    print('Could not reached desired value. Went to maximum power %fuW'%actual_power)
                #if actual_posi <= limits_filterwheel[0]+step_size:
                #    print('Reached upper limit of filterwheel position, power is %fuW'%actual_power)
                #    return actual_power
            print('reached target: %fuW.'%actual_power)
            return actual_power
        else:
            print('power was fine: %fuW'%actual_power)
            return actual_power
    
ttd = talk_to_devices()
```
