```python
import pandas as pd
import numpy as np
import glob
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
```

# Claibration spectrometer

```python
# TODO: calibrate spectrometer
```

```python
calib = np.genfromtxt(r"Z:\Oliver\measurements\238\2025-03-21-PL-spectra\calibration-spectrometer.txt")
```

```python
def func(xx, a, b):
    return a*xx+b

wl_laser = calib[:,0]
wl_spec = calib[:,1]
XX = np.linspace(min(wl_spec), max(wl_spec),100)

popt, pcov = curve_fit(func, wl_spec, wl_laser)

plt.plot(wl_spec, wl_laser, 'o', label='data')
plt.plot(XX, func(XX,*popt), label = '%.2f*x+%.2f'%(popt[0], popt[1]))
plt.legend()
plt.xlabel('lambda spec')
plt.ylabel('lambda laser')
```

# 60keV (not yet measured)

## soft-ICP

```python
# Get all .dat files from the specified folder
dat_files = glob.glob(r'Z:\Oliver\measurements\238\2025-03-21-PL-spectra\D_60keV_soft-icp\*.dat')
title_str = '60keV, soft-icp'
minpeak = 400

# Initialize an empty list to store the data
data_list = []

# Loop through each file and read the data
for file in dat_files:
    # Read the data from the file, skipping lines that start with '#'
    wavelength, data = np.genfromtxt(file, comments='#', usecols=(0, 1), unpack=True)
    # Check if there are peaks with intensity larger than 800
    if np.any(data > minpeak):
        data_list.append(data)
wavelength = wavelength/1e9
wavelength = func(wavelength,*popt)


# Stack the data vertically to create a 2D numpy array
if data_list:
    data_array = np.vstack(data_list)

    # Plot the data using imshow
    plt.imshow(data_array, aspect='auto', extent=[wavelength.min(), wavelength.max(), 0, len(data_list)], vmax=2000)
    plt.xlabel('Wavelength (nm)')
    plt.ylabel('File Index')
    plt.colorbar(label='Intensity')
    plt.title(title_str)
    plt.show()
else:
    print("No data with peaks larger than %d found."%minpeak)
```

```python
# Stack the data vertically to create a 2D numpy array
if data_list:
    data_array = np.vstack(data_list)

    # Define the wavelength range for zooming
    zoom_start = 618  # 620 nm
    zoom_end = 621    # 622 nm

    # Create a mask for the zoomed-in range
    mask = (wavelength >= zoom_start) & (wavelength <= zoom_end)

    # Plot the data using imshow with the zoomed-in range
    plt.imshow(data_array[:, mask], aspect='auto', extent=[zoom_start, zoom_end, 0, len(data_list)], vmax=2000, interpolation='nearest')
    plt.xlabel('Wavelength (nm)')
    plt.ylabel('File Index')
    plt.colorbar(label='Intensity')
    plt.title(title_str)
    plt.xlim(zoom_start, zoom_end)
    plt.show()
else:
    print("No data with peaks larger than 800 found.")
```
