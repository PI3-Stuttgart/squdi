```python
import serial
import time

# Set up the serial connection
ser = serial.Serial(
    port='COM3',      # Replace with your COM port
    baudrate=9600,    # Match the spectrometer's baud rate
    bytesize=serial.EIGHTBITS,
    parity=serial.PARITY_NONE,
    stopbits=serial.STOPBITS_ONE,
    timeout=2         # Timeout for reading responses
)

def send_command(command):
    """Send a command to the spectrometer and wait for the response."""
    ser.write((command + '\r').encode())  # Send the command with CR
    time.sleep(0.1)  # Short delay to allow command processing
    
    response = []
    while True:
        line = ser.readline().decode().strip()  # Read each line
        if not line:  # Stop reading when no more data
            break
        response.append(line)
    
    return '\n'.join(response)



def wait_for_ok(timeout=5):
    """Wait specifically for the 'OK' response from the spectrometer with a timeout."""
    start_time = time.time()  # Record the start time
    while True:
        # Check if timeout has been reached
        if time.time() - start_time > timeout:
            raise TimeoutError("Timeout while waiting for 'OK' response from the spectrometer.")

        response = ser.readline().decode().strip()  # Read the response
        if response == "OK":
            break  # Exit loop when 'OK' is received
        elif response:  # Log any unexpected output
            print(f"Unexpected response: {response}")

def set_grating(grating_number: int):
    """
    Set the specified grating by its number.
    
    Parameters:
        grating_number (int): The number of the grating to set (1-9).
    
    Returns:
        str: Response from the spectrometer.
    """
    if not 1 <= grating_number <= 9:
        raise ValueError("Grating number must be between 1 and 9.")
    return send_command(f"{grating_number} GRATING")
def get_current_grating():
    """
    Get the currently selected grating.
    
    Returns:
        str: Current grating number.
    """
    return send_command("?GRATING")

def get_current_turret():
    """
    Get the currently selected grating.
    
    Returns:
        str: Current grating number.
    """
    return send_command("?TURRET")
def set_turret(turret: int):
    """
    Set the specified grating by its number.
    
    Parameters:
        grating_number (int): The number of the turret to set (1-3).
    
    Returns:
        str: Response from the spectrometer.
    """
    if not 1 <= turret <= 9:
        raise ValueError("Grating number must be between 1 and 9.")
    return send_command(f"{turret} GRATING")
def get_grating_info():
    """
    Get detailed information about all installed gratings.
    
    Returns:
        str: List of installed gratings with details.
    """
    return send_command("?GRATINGS")
def set_wavelength(wavelength: float):
    """
    Set the monochromator to a specific wavelength.
    
    Parameters:
        wavelength (float): The target wavelength in nanometers.
    
    Returns:
        str: Response from the spectrometer.
    """
    if wavelength < 0 or wavelength > 1400:
        raise ValueError("Wavelength must be between 0 and 1400 nm.")
    return send_command(f"{wavelength:.2f} GOTO")
def get_current_wavelength():
    """
    Get the current wavelength of the monochromator.
    
    Returns:
        str: Current wavelength in nanometers.
    """
    return send_command("?NM")
```
