```python
import serial
import time

# Set up the serial connection
ser = serial.Serial(
    port='COM3',      # Replace with your COM port
    baudrate=9600,    # Match the spectrometer's baud rate
    bytesize=serial.EIGHTBITS,
    parity=serial.PARITY_NONE,
    stopbits=serial.STOPBITS_ONE,
    timeout=2         # Timeout for reading responses
)

def send_command(command):
    """Send a command to the spectrometer and wait for the response."""
    ser.write((command + '\r').encode())  # Send the command with CR
    time.sleep(0.1)  # Short delay to allow command processing
    
    response = []
    while True:
        line = ser.readline().decode().strip()  # Read each line
        if not line:  # Stop reading when no more data
            break
        response.append(line)
    
    return '\n'.join(response)



def wait_for_ok(timeout=5):
    """Wait specifically for the 'OK' response from the spectrometer with a timeout."""
    start_time = time.time()  # Record the start time
    while True:
        # Check if timeout has been reached
        if time.time() - start_time > timeout:
            raise TimeoutError("Timeout while waiting for 'OK' response from the spectrometer.")

        response = ser.readline().decode().strip()  # Read the response
        if response == "OK":
            break  # Exit loop when 'OK' is received
        elif response:  # Log any unexpected output
            print(f"Unexpected response: {response}")

def set_grating(grating_number: int):
    """
    Set the specified grating by its number.
    
    Parameters:
        grating_number (int): The number of the grating to set (1-9).
    
    Returns:
        str: Response from the spectrometer.
    """
    if not 1 <= grating_number <= 9:
        raise ValueError("Grating number must be between 1 and 9.")
    return send_command(f"{grating_number} GRATING")
def get_current_grating():
    """
    Get the currently selected grating.
    
    Returns:
        str: Current grating number.
    """
    return send_command("?GRATING")

def get_current_turret():
    """
    Get the currently selected grating.
    
    Returns:
        str: Current grating number.
    """
    return send_command("?TURRET")
def set_turret(turret: int):
    """
    Set the specified grating by its number.
    
    Parameters:
        grating_number (int): The number of the turret to set (1-3).
    
    Returns:
        str: Response from the spectrometer.
    """
    if not 1 <= turret <= 9:
        raise ValueError("Grating number must be between 1 and 9.")
    return send_command(f"{turret} GRATING")
def get_grating_info():
    """
    Get detailed information about all installed gratings.
    
    Returns:
        str: List of installed gratings with details.
    """
    return send_command("?GRATINGS")
def set_wavelength(wavelength: float):
    """
    Set the monochromator to a specific wavelength.
    
    Parameters:
        wavelength (float): The target wavelength in nanometers.
    
    Returns:
        str: Response from the spectrometer.
    """
    if wavelength < 0 or wavelength > 1400:
        raise ValueError("Wavelength must be between 0 and 1400 nm.")
    return send_command(f"{wavelength:.2f} GOTO")
def get_current_wavelength():
    """
    Get the current wavelength of the monochromator.
    
    Returns:
        str: Current wavelength in nanometers.
    """
    return send_command("?NM")
```

```python
import ctypes
from ctypes import c_void_p, c_int, POINTER, byref

class PICam:
    def __init__(self, dll_path: str = "pvcam64.dll"):
        """Initialize the PICam API and load the DLL."""
        self.dll = ctypes.windll.LoadLibrary(dll_path)
        self.camera_handle = c_void_p()
        self.check_error(self.dll.Picam_InitializeLibrary())
        print("PICam Library Initialized")

    def check_error(self, error_code):
        """Check for errors returned by PICam functions."""
        if error_code != 0:  # Assuming 0 means success
            raise Exception(f"PICam Error: {error_code}")

    def get_camera_handle(self):
        """Discover and open the first available camera."""
        # Get list of available cameras
        camera_id_array = POINTER(c_void_p)()
        camera_count = c_int()

        self.check_error(self.dll.Picam_GetAvailableCameraIDs(
            byref(camera_id_array), byref(camera_count)
        ))

        if camera_count.value == 0:
            raise Exception("No cameras found")

        # Open the first camera
        self.check_error(self.dll.Picam_OpenCamera(
            camera_id_array[0], byref(self.camera_handle)
        ))
        print("Camera opened successfully!")

        # Cleanup camera IDs
        self.check_error(self.dll.Picam_DestroyCameraIDs(camera_id_array))

    def close_camera(self):
        """Close the camera handle."""
        if self.camera_handle:
            self.check_error(self.dll.Picam_CloseCamera(self.camera_handle))
            print("Camera closed successfully!")

    def uninitialize_library(self):
        """Uninitialize the PICam library."""
        self.check_error(self.dll.Picam_UninitializeLibrary())
        print("PICam Library Uninitialized")
```

```python
# Step 1: Import Necessary Libraries
import ctypes
from ctypes import c_int, c_double, c_void_p, POINTER, byref
import numpy as np
import matplotlib.pyplot as plt
from ctypes import Structure, POINTER, c_void_p, c_longlong

class PicamAvailableData(Structure):
    _fields_ = [
        ("initial_readout", c_void_p),
        ("readout_count", c_longlong),
    ]

class PicamCameraID(ctypes.Structure):
    _fields_ = [
        ("model", c_int),
        ("computer_interface", c_int),
        ("sensor_name", ctypes.c_char * 64),
        ("serial_number", ctypes.c_char * 64),
    ]
class PicamAcquisitionErrorsMask(Structure):
    _fields_ = [("mask", c_int)]
# Step 2: Define the Picam Wrapper Class
class PICam:
    def __init__(self, dll_path: str):
        """Initialize the PICam API and load the DLL."""
        self.dll = ctypes.windll.LoadLibrary(dll_path)
        self.camera_handle = c_void_p()

    def get_camera_ids(self):
        """Get available camera IDs."""
        id_array = POINTER(PicamCameraID)()
        id_count = c_int()
        self.check_error(self.dll.Picam_GetAvailableCameraIDs(byref(id_array), byref(id_count)))
        return id_array, id_count.value
        
    def initialize(self):
        """Initialize the library and ensure it's ready."""
        result = self.dll.Picam_InitializeLibrary()
        if result != 0:
            raise Exception(f"Failed to initialize PICam library. Error code: {result}")
        print("PICam library initialized successfully.")
        
    def open_camera(self):
        """Open the first available camera."""
        # Define the camera list pointer and count
        camera_list = POINTER(PicamCameraID)()
        camera_count = c_int()
        
        # Retrieve available camera IDs
        result = self.dll.Picam_GetAvailableCameraIDs(byref(camera_list), byref(camera_count))
        if result != 0:
            raise Exception(f"Failed to get camera IDs. Error code: {result}")
        
        if camera_count.value == 0:
            raise Exception("No cameras found.")
        
        print(f"Found {camera_count.value} camera(s). Opening the first one.")
        
        # Access the first camera in the list
        first_camera = camera_list[0]
        
        # Open the camera
        self.check_error(self.dll.Picam_OpenCamera(byref(first_camera), byref(self.camera_handle)))
        print("Camera opened successfully.")

    def __start_acquisition(self):
        """Perform minimal acquisition test."""
        readout_count = 1
        timeout_ms = 100
        available_data = PicamAvailableData()
        errors = PicamAcquisitionErrorsMask()
        
        print("Attempting to acquire data...")
        result = self.dll.Picam_Acquire(
            self.camera_handle,
            c_longlong(readout_count),
            c_int(timeout_ms),
            byref(available_data),
            byref(errors),
        )
        if result != 0:
            raise Exception(f"Failed to acquire data. Error code: {result}")
        
        print(f"Acquisition successful! Readout Count: {available_data.readout_count}")
        return available_data

    def start_acquisition(self, readout_count=1, timeout_ms=100):
        
        

        available_data = PicamAvailableData()
        errors = c_int()
        result = self.dll.Picam_Acquire(
            self.camera_handle,
            c_longlong(readout_count),
            c_int(timeout_ms),
            byref(available_data),
            byref(errors),
        )
        if result != 0:
            error_buffer = create_string_buffer(256)
            self.dll.Picam_GetExtendedError(result, error_buffer, len(error_buffer))
            raise Exception(f"Acquisition failed. Error {result}: {error_buffer.value.decode('utf-8')}")
        print("Acquisition successful!")
        return available_data
    

    def stop_acquisition(self):
        """Stop the camera acquisition process."""
        self.check_error(self.dll.Picam_StopAcquisition(self.camera_handle))
        print("Acquisition stopped")

    def acquire_data(self, readout_count=1, timeout_ms=100):
        """Acquire data from the camera."""
        # Check if acquisition is running and stop if necessary
        if self.is_acquisition_running():
            print("Acquisition already in progress. Stopping it now.")
            self.stop_acquisition()
    
        available_data = PicamAvailableData()
        errors = c_int()
        self.check_error(self.dll.Picam_Acquire(
            self.camera_handle,
            c_longlong(readout_count),
            c_int(timeout_ms),
            byref(available_data),
            byref(errors),
        ))
        print(f"Readout Count: {available_data.readout_count}")
        return available_data

    def is_acquisition_running(self):
        """Check if an acquisition is currently running."""
        running = c_int()
        self.check_error(self.dll.Picam_IsAcquisitionRunning(self.camera_handle, byref(running)))
        return bool(running.value)
    def stop_acquisition(self):
        """Stop the current acquisition."""
        self.check_error(self.dll.Picam_StopAcquisition(self.camera_handle))
        print("Acquisition stopped successfully.")

    
    def close_camera(self):
        """Close the camera handle and uninitialize the library."""
        if self.camera_handle:
            self.dll.Picam_CloseCamera(self.camera_handle)
            print("Camera closed successfully.")
        
        self.dll.Picam_UninitializeLibrary()
        print("PICam library uninitialized successfully.")

    def check_error(self, result):
        """Check for errors and raise exceptions."""
        if result != 0:
            raise Exception(f"PICam Error: {result}")

    def uninitialize(self):
        """Uninitialize the PICam library."""
        self.check_error(self.dll.Picam_UninitializeLibrary())
        print("PICam Library Uninitialized")
    def reset_camera(self):
        """Reset the camera and library."""
        self.stop_acquisition()
        self.check_error(self.dll.Picam_CloseCamera(self.camera_handle))
        self.check_error(self.dll.Picam_UninitializeLibrary())
        print("Camera and library reset successfully.")
```

```python
from ctypes import *

class Camera:
    def __init__(self):
        self.handle = c_int16()
        self.dll = windll.LoadLibrary("C:\\WINDOWS\\system32\\pvcam64.dll")
    
    def open_camera(self):
        """Open the camera."""
        self.initialize_library()
        
        cam_name = create_string_buffer(255)
        self.check_error(self.dll.pl_cam_get_name(0, cam_name))
        self.check_error(self.dll.pl_cam_open(cam_name, byref(self.handle), OPEN_EXCLUSIVE))
        print("Camera opened successfully.")
    def initialize_library(self):
        """Initialize the PVCAM library."""
        result = self.dll.pl_pvcam_init()
        if not result:
            code, message = self.get_error()
            raise Exception(f"Failed to initialize library: {message}")
        print("PVCAM library initialized successfully.")
    
        
    def acquire_data(self, exp_time=1000):
        """Acquire data from the camera."""
        try:
            self.initialize_sequence()
            self.set_exposure_time(exp_time)
            self.start_sequence()
            return self.retrieve_data()
        except Exception as e:
            print(f"Acquisition failed: {e}")
            self.reset_camera()
            raise

    def set_exposure_time(self, exp_time):
        """Set the exposure time."""
        self.set_param(PARAM_EXP_TIME, exp_time)
        print(f"Exposure time set to {exp_time} µs.")

    def start_sequence(self):
        """Start the acquisition sequence."""
        size = c_uint32()
        self.check_error(self.dll.pl_exp_setup_seq(
            self.handle, 1, 1, byref(self.region), TIMED_MODE, c_uint32(self.exp_time), byref(size)
        ))
        print("Sequence started.")

    def retrieve_data(self):
        """Retrieve data from the acquisition."""
        status = c_int16(READOUT_NOT_ACTIVE)
        dummy = c_uint32()
        while status.value != READOUT_COMPLETE:
            self.check_error(self.dll.pl_exp_check_status(self.handle, byref(status), byref(dummy)))
        print("Data acquisition complete.")
        return self.frames

    def close_camera(self):
        """Close the camera."""
        self.check_error(self.dll.pl_cam_close(self.handle))
        self.dll.pl_pvcam_uninit()
        print("Camera closed and uninitialized.")

    def check_error(self, result):
        """Check for errors and raise exceptions."""
        if not result:
            code, message = self.get_error()
            raise Exception(f"PICam Error {code}: {message}")

    def get_error(self):
        """Retrieve the last error code and message."""
        code = self.dll.pl_error_code()
        message = create_string_buffer(255)
        self.dll.pl_error_message(code, message)
        return code, message.value
```

```python
import ctypes
from ctypes import c_int, c_double, c_void_p, byref, POINTER, Structure


class PicamCameraID(ctypes.Structure):
    _fields_ = [
        ("model", c_int),
        ("computer_interface", c_int),
        ("sensor_name", ctypes.c_char * 64),
        ("serial_number", ctypes.c_char * 64),
    ]


class PicamAcquisitionErrorsMask(Structure):
    _fields_ = [("mask", c_int)]


# class PicamAcquiredReadout(ctypes.Structure):
#     _fields_ = [
#         ("timestamp", ctypes.c_double),  # Example field for timestamp
#         ("data", ctypes.POINTER(ctypes.c_double)),  # Example field for data pointer
#         ("data_count", ctypes.c_int),  # Example field for count of data points
#     ]
# class PicamReadout(ctypes.Structure):
#     _fields_ = [
#         ("frame_data", ctypes.POINTER(ctypes.c_uint16)),  # Adjust type based on your data
#         ("frame_size", ctypes.c_int),
#     ]
class PicamReadout(ctypes.Structure):
    _fields_ = [
        ("data", ctypes.POINTER(ctypes.c_int64)),  # Assuming data is 16-bit unsigned integers
        ("frame_size", ctypes.c_int),  # Use c_size_t for size-related fields
    ]

# class PicamReadout(ctypes.Structure):
#     _fields_ = [
#         ("data", ctypes.POINTER(ctypes.c_double)),  # Example field
#         ("timestamp", ctypes.c_double),  # Example field
#     ]
    
class PicamAvailableData(Structure):
    _fields_ = [("initial_readout", c_void_p), ("readout_count", ctypes.c_longlong)]



class SpectrometerController:
    def __init__(self, library_path):
        """
        Initialize the spectrometer controller.

        :param library_path: Path to the PICam shared library (DLL or .so file).
        """
        self.dll = ctypes.cdll.LoadLibrary(library_path)
        self.camera_handle = c_void_p()

    def get_frame_size(self):
        frame_size = ctypes.c_int()
        self.check_error(
            self.dll.Picam_GetParameterIntegerValue(
                self.camera_handle,
                ctypes.c_int(PicamParameter_FrameSize),
                ctypes.byref(frame_size)
            )
        )
        return frame_size.value
    
    def check_error(self, error_code):
        """
        Check and raise an exception if the error code is not PicamError_None.

        :param error_code: Error code returned by a PICam function.
        """
        if error_code != 0:  # PicamError_None = 0
            raise Exception(f"PICam Error {error_code}: {self.get_error_string(error_code)}")

    def get_error_string(self, error_code):
        """
        Get a string description of the given error code.

        :param error_code: Error code returned by a PICam function.
        :return: Description of the error.
        """
        self.dll.Picam_GetEnumerationString.restype = ctypes.POINTER(ctypes.c_char_p)
        error_string = ctypes.c_char_p()
        self.dll.Picam_GetEnumerationString(c_int(1), error_code, byref(error_string))  # 1 = PicamEnumeratedType_Error
        return error_string.value.decode()

    def initialize(self):
        """
        Initialize the PICam library.
        """
        self.check_error(self.dll.Picam_InitializeLibrary())

    def uninitialize(self):
        """
        Uninitialize the PICam library.
        """
        self.check_error(self.dll.Picam_UninitializeLibrary())

    def list_cameras(self):
        """
        List available cameras.

        :return: List of PicamCameraID instances.
        """
        camera_ids = POINTER(PicamCameraID)()
        camera_count = c_int()
        self.check_error(self.dll.Picam_GetAvailableCameraIDs(byref(camera_ids), byref(camera_count)))

        cameras = []
        for i in range(camera_count.value):
            cameras.append(camera_ids[i])

        self.dll.Picam_DestroyCameraIDs(camera_ids)
        return cameras

    def open_camera(self, camera_id):
        """
        Open a specific camera by its ID.

        :param camera_id: An instance of PicamCameraID.
        """
        self.check_error(self.dll.Picam_OpenCamera(byref(camera_id), byref(self.camera_handle)))

    def close_camera(self):
        """
        Close the currently opened camera.
        """
        self.check_error(self.dll.Picam_CloseCamera(self.camera_handle))
        self.camera_handle = None

    def acquire_data(self, readout_count=1, timeout_ms=1000):
        """
        Acquire data from the spectrometer.

        :param readout_count: Number of readouts to acquire.
        :param timeout_ms: Timeout in milliseconds for acquisition.
        :return: The acquired data and any acquisition errors as PicamAvailableData and PicamAcquisitionErrorsMask.
        """
        available_data = PicamAvailableData()
        errors = PicamAcquisitionErrorsMask()

        # Allocate space for acquired readouts
        readouts = (PicamReadout * readout_count)()  # Array of structures
    
        # Update `available_data` to point to this array
        available_data.initial_readout = ctypes.cast(readouts, c_void_p)
        
        self.check_error(
            self.dll.Picam_Acquire(
                self.camera_handle,
                c_int(readout_count),
                c_int(timeout_ms),
                byref(available_data),
                byref(errors),
            )
        )
        return available_data, errors

    def start_acquisition(self):
        """
        Start data acquisition.
        """
        self.check_error(self.dll.Picam_StartAcquisition(self.camera_handle))

    def stop_acquisition(self):
        """
        Stop data acquisition.
        """
        self.check_error(self.dll.Picam_StopAcquisition(self.camera_handle))

    def is_acquisition_running(self):
        """
        Check if acquisition is running.

        :return: True if acquisition is running, False otherwise.
        """
        running = c_int()
        self.check_error(self.dll.Picam_IsAcquisitionRunning(self.camera_handle, byref(running)))
        return bool(running.value)

    def set_parameter(self, parameter, value):
        """
        Set a camera parameter.

        :param parameter: Parameter ID to set.
        :param value: Value to set for the parameter.
        """
        if isinstance(value, int):
            self.check_error(self.dll.Picam_SetParameterIntegerValue(self.camera_handle, parameter, c_int(value)))
        elif isinstance(value, float):
            self.check_error(self.dll.Picam_SetParameterFloatingPointValue(self.camera_handle, parameter, c_double(value)))
        else:
            raise TypeError("Unsupported parameter value type. Must be int or float.")

    def get_parameter(self, parameter):
        """
        Get the value of a camera parameter.

        :param parameter: Parameter ID to retrieve.
        :return: Value of the parameter.
        """
        value = c_double()
        self.check_error(self.dll.Picam_GetParameterFloatingPointValue(self.camera_handle, parameter, byref(value)))
        return value.value
    def get_parameter_value_type(self, parameter_id):
        """Get the value type of a parameter."""
        value_type = ctypes.c_int()
        self.check_error(
            self.dll.Picam_GetParameterValueType(
                self.camera_handle,
                ctypes.c_int(parameter_id),
                ctypes.byref(value_type)
            )
        )
        return value_type.value

        
    def process_available_data(self, available_data):
        """
        Process the available data from the acquisition.
    
        :param available_data: PicamAvailableData structure containing the acquisition results.
        :return: List of processed frames.
        """
        readout_count = available_data.readout_count
        initial_readout = available_data.initial_readout
    
        if readout_count == 0 or not initial_readout:
            raise Exception("No data available.")
    
        # Cast initial_readout to an array of PicamReadout structures
        readouts = ctypes.cast(
            initial_readout, ctypes.POINTER(PicamReadout * readout_count)
        ).contents
    
        # Process each readout
        processed_frames = []
        for i in range(readout_count):
            readout = readouts[i]
    
            # Convert frame data to a Python list or NumPy array
            frame = (ctypes.c_uint16 * readout.frame_size).from_address(
                ctypes.addressof(readout.frame_data.contents)
            )
    
            # Example: Convert to a Python list
            processed_frames.append(list(frame))
    
        return processed_frames
```

```python
import ctypes
from ctypes import c_int, c_double, c_void_p, byref, POINTER, Structure, c_longlong


class PicamCameraID(ctypes.Structure):
    _fields_ = [
        ("model", c_int),
        ("computer_interface", c_int),
        ("sensor_name", ctypes.c_char * 64),
        ("serial_number", ctypes.c_char * 64),
    ]


class PicamAcquisitionErrorsMask(Structure):
    _fields_ = [("mask", c_int)]


class PicamReadout(ctypes.Structure):
    _fields_ = [
        
        ("frame_size", c_int),  # Size of the data frame
        ("data", POINTER(ctypes.c_int64)),  # Pointer to the readout data
    ]


class PicamAvailableData(Structure):
    _fields_ = [("initial_readout", c_void_p), ("readout_count", c_longlong)]


class SpectrometerController:
    def __init__(self, library_path):
        """
        Initialize the spectrometer controller.

        :param library_path: Path to the PICam shared library (DLL or .so file).
        """
        self.dll = ctypes.cdll.LoadLibrary(library_path)
        self.camera_handle = c_void_p()

    def check_error(self, error_code):
        """
        Check and raise an exception if the error code is not PicamError_None.

        :param error_code: Error code returned by a PICam function.
        """
        if error_code != 0:  # PicamError_None = 0
            raise Exception(f"PICam Error {error_code}: {self.get_error_string(error_code)}")

    def get_error_string(self, error_code):
        """
        Get a string description of the given error code.

        :param error_code: Error code returned by a PICam function.
        :return: Description of the error.
        """
        self.dll.Picam_GetEnumerationString.restype = ctypes.POINTER(ctypes.c_char_p)
        error_string = ctypes.c_char_p()
        self.dll.Picam_GetEnumerationString(c_int(1), error_code, byref(error_string))  # 1 = PicamEnumeratedType_Error
        return error_string.value.decode()

    def initialize(self):
        """Initialize the PICam library."""
        self.check_error(self.dll.Picam_InitializeLibrary())

    def uninitialize(self):
        """Uninitialize the PICam library."""
        self.check_error(self.dll.Picam_UninitializeLibrary())

    def open_camera(self, camera_id):
        """
        Open a specific camera by its ID.

        :param camera_id: An instance of PicamCameraID.
        """
        self.check_error(self.dll.Picam_OpenCamera(byref(camera_id), byref(self.camera_handle)))

    def close_camera(self):
        """Close the currently opened camera."""
        self.check_error(self.dll.Picam_CloseCamera(self.camera_handle))
        self.camera_handle = None

    def acquire_data(self, required_frame_size, readout_count=1, timeout_ms=1000):
        """
        Acquire data from the spectrometer and ensure frame size matches the requirement.

        :param required_frame_size: Expected frame size for the acquisition.
        :param readout_count: Number of readouts to acquire.
        :param timeout_ms: Timeout in milliseconds for acquisition.
        :return: List of processed frames.
        """
        available_data = PicamAvailableData()
        errors = PicamAcquisitionErrorsMask()

        # Allocate space for acquired readouts
        readouts = (PicamReadout * readout_count)()  # Array of structures
        available_data.initial_readout = ctypes.cast(readouts, c_void_p)

        self.check_error(
            self.dll.Picam_Acquire(
                self.camera_handle,
                c_int(readout_count),
                c_int(timeout_ms),
                byref(available_data),
                byref(errors),
            )
        )

        # Validate readout frame sizes
        for i in range(available_data.readout_count):
            if readouts[i].frame_size != required_frame_size:
                raise ValueError(
                    f"Frame size mismatch: Expected {required_frame_size}, got {readouts[i].frame_size}"
                )

        return self.process_available_data(readouts, available_data.readout_count)
    def start_acquisition(self):
        """
        Start data acquisition.
        """
        self.check_error(self.dll.Picam_StartAcquisition(self.camera_handle))

    def stop_acquisition(self):
        """
        Stop data acquisition.
        """
        self.check_error(self.dll.Picam_StopAcquisition(self.camera_handle))

    def is_acquisition_running(self):
        """
        Check if acquisition is running.

        :return: True if acquisition is running, False otherwise.
        """
        running = c_int()
        self.check_error(self.dll.Picam_IsAcquisitionRunning(self.camera_handle, byref(running)))
        return bool(running.value)
    def list_cameras(self):
        """
        List available cameras.

        :return: List of PicamCameraID instances.
        """
        camera_ids = POINTER(PicamCameraID)()
        camera_count = c_int()
        self.check_error(self.dll.Picam_GetAvailableCameraIDs(byref(camera_ids), byref(camera_count)))

        cameras = []
        for i in range(camera_count.value):
            cameras.append(camera_ids[i])

        self.dll.Picam_DestroyCameraIDs(camera_ids)
        return cameras
        
    def process_available_data(self, readouts, readout_count):
        """
        Process the available data from the acquisition.

        :param readouts: Array of PicamReadout structures.
        :param readout_count: Number of readouts to process.
        :return: List of processed frames.
        """
        processed_frames = []
        for i in range(readout_count):
            readout = readouts[i]
            frame = (ctypes.c_int64 * readout.frame_size).from_address(
                ctypes.addressof(readout.data.contents)
            )
            processed_frames.append(list(frame))  # Convert to a Python list
        return processed_frames

    def set_parameter(self, parameter, value):
        """
        Set a camera parameter.

        :param parameter: Parameter ID to set.
        :param value: Value to set for the parameter.
        """
        if isinstance(value, int):
            self.check_error(self.dll.Picam_SetParameterIntegerValue(self.camera_handle, parameter, c_int(value)))
        elif isinstance(value, float):
            self.check_error(self.dll.Picam_SetParameterFloatingPointValue(self.camera_handle, parameter, c_double(value)))
        else:
            raise TypeError("Unsupported parameter value type. Must be int or float.")

    def get_parameter(self, parameter):
        """
        Get the value of a camera parameter.

        :param parameter: Parameter ID to retrieve.
        :return: Value of the parameter.
        """
        value = c_double()
        self.check_error(self.dll.Picam_GetParameterFloatingPointValue(self.camera_handle, parameter, byref(value)))
        return value.value
```

```python
def get_error_string(error_code):
    """
    Get a string description of the given error code.

    :param error_code: Error code returned by a PICam function.
    :return: Description of the error.
    """
    self.dll.Picam_GetEnumerationString.restype = ctypes.POINTER(ctypes.c_char_p)
    error_string = ctypes.c_char_p()
    self.dll.Picam_GetEnumerationString(c_int(1), error_code, byref(error_string))  # 1 = PicamEnumeratedType_Error
    return error_string.value.decode()
```

```python
from ctypes import POINTER, c_long, c_int, byref

class CameraParameterManager:
    def __init__(self, controller):
        """
        Initializes the CameraParameterManager with the provided controller.
        
        :param controller: The SpectrometerController instance with DLL and camera handle.
        """
        self.controller = controller
        self.parameters = {}
        self.parameter_count = 0

    def load_parameters(self):
        """
        Loads all supported parameters for the connected camera.
        """
        parameter_count = c_int()
        parameters = POINTER(c_long)()

        # Call the Picam_GetParameters function
        error_code = self.controller.dll.Picam_GetParameters(
            self.controller.camera_handle,
            byref(parameters),
            byref(parameter_count)
        )
        self.controller.check_error(error_code)

        # Save parameters to a dictionary mapping ID to function number
        self.parameters = {
            parameters[i]: self.get_function_number(parameters[i])
            for i in range(parameter_count.value)
        }
        self.parameter_count = parameter_count.value
        print("Loaded parameters successfully.")

    @staticmethod
    def get_function_number(parameter_id):
        """
        Extracts the function number (n) from a PicamParameter ID.

        :param parameter_id: The parameter ID as an integer (hexadecimal value in numeric form).
        :return: The function number (n).
        """
        return parameter_id & 0xFFFF  # Mask the lower 16 bits

    def is_parameter_supported(self, parameter_id):
        """
        Checks if a specific parameter is supported by the camera.

        :param parameter_id: The parameter ID to check.
        :return: True if the parameter is supported, False otherwise.
        """
        return parameter_id in self.parameters

    def query_parameter(self, parameter_id):
        """
        Queries the camera for a specific parameter and prints its function number if supported.

        :param parameter_id: The parameter ID to query.
        :return: The function number if supported, or a message if not.
        """
        if self.is_parameter_supported(parameter_id):
            return f"Parameter ID: {hex(parameter_id)} is supported. Function Number: {self.parameters[parameter_id]}"
        else:
            return f"Parameter ID: {hex(parameter_id)} is not supported by the camera."

    def print_all_parameters(self):
        """
        Prints all available parameters and their function numbers.
        """
        print("Supported Parameters and Function Numbers:")
        for param_id, func_num in self.parameters.items():
            print(f"Parameter ID: {hex(param_id)}, Function Number: {func_num}")
```

```python
# Example usage
# controller = ...  # Initialize your SpectrometerController instance here

# Initialize the parameter manager
param_manager = CameraParameterManager(controller)

# Load and print all parameters
param_manager.load_parameters()
param_manager.print_all_parameters()

# Query a specific parameter
param_id_to_query = 0x02170017  # Example parameter
print(param_manager.query_parameter(param_id_to_query))
```

```python
from ctypes import POINTER, c_int, c_double, byref, c_long, c_bool

def get_parameter_value(controller, camera_handle, parameter_id):
    """
    Retrieve the value of a specified parameter from the camera.

    :param controller: An instance of the camera controller with the DLL loaded.
    :param camera_handle: Handle to the connected camera.
    :param parameter_id: The parameter ID to query.
    :return: The parameter value.
    """
    # Get the parameter type
    param_type = c_int()
    error_code = controller.dll.Picam_GetParameterValueType(
        camera_handle,
        c_long(parameter_id),
        byref(param_type)
    )
    controller.check_error(error_code)

    # Check parameter type and retrieve the value
    if param_type.value == 1:  # Integer
        value = c_int()
        error_code = controller.dll.Picam_GetParameterIntegerValue(
            camera_handle,
            c_long(parameter_id),
            byref(value)
        )
        controller.check_error(error_code)
        return value.value

    elif param_type.value == 2:  # FloatingPoint
        value = c_double()
        error_code = controller.dll.Picam_GetParameterFloatingPointValue(
            camera_handle,
            c_long(parameter_id),
            byref(value)
        )
        controller.check_error(error_code)
        return value.value

    elif param_type.value == 3:  # Boolean
        value = c_bool()
        error_code = controller.dll.Picam_GetParameterIntegerValue(
            camera_handle,
            c_long(parameter_id),
            byref(value)
        )
        controller.check_error(error_code)
        return value.value

    elif param_type.value == 4:  # Enumeration
        value = c_int()
        error_code = controller.dll.Picam_GetParameterIntegerValue(
            camera_handle,
            c_long(parameter_id),
            byref(value)
        )
        controller.check_error(error_code)
        return value.value

    elif param_type.value == 6:  # Large Integer
        value = c_long()
        error_code = controller.dll.Picam_GetParameterLargeIntegerValue(
            camera_handle,
            c_long(parameter_id),
            byref(value)
        )
        controller.check_error(error_code)
        return value.value

    else:
        raise Exception(f"Unsupported parameter type: {param_type.value}")
```
