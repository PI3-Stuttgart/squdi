```python
from PyQt5.QtCore import QTimer

def print_message():
    print("This message is printed every 0.2 seconds")

def stop_timer():
    timer.stop()
    print("Timer stopped")

# Create a QTimer instance
timer = QTimer()

# Connect the timer's timeout signal to the print_message function
timer.timeout.connect(print_message)

# Set the timer interval to 200 milliseconds (0.2 seconds)
timer.start(200)
```

```python
stop_timer()
```
