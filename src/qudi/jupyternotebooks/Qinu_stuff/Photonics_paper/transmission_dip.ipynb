```python
def waveguide_angle2lab_angle(waveguide_angle, TE_transmission = 130, TM_transmission=15):
    return np.rad2deg(np.arctan(np.tan(np.deg2rad(waveguide_angle)) * np.sqrt(TE_transmission) / np.sqrt(TM_transmission) ))
```

```python
def lab_angle2waveguide_angle(lab_angle, TE_transmission = 130, TM_transmission=15):
    return np.rad2deg(np.arctan(np.tan(np.deg2rad(lab_angle)) * np.sqrt(TM_transmission) / np.sqrt(TE_transmission) ))
```
