```python
import numpy as np

# Given values
NA_1 = 0.95   # Numerical aperture for first objective
NA_2 = 0.9    # Numerical aperture for second objective
n_diamond = 2.4  # Refractive index of diamond
counts_na_1 = 30000  # Counts per second for NA = 0.95

# Calculate collection efficiency for both NAs
theta_1 = np.arcsin(NA_1 / n_diamond)
theta_2 = np.arcsin(NA_2 / n_diamond)

eta_1 = 1 - np.cos(theta_1)
eta_2 = 1 - np.cos(theta_2)

# Calculate the expected counts for the second NA
counts_na_2 = counts_na_1 * (eta_2 / eta_1)
counts_na_2
```
