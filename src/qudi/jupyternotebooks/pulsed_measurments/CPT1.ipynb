```python
import serial
import time

# --- CONFIGURATION ---
# Replace 'COM3' with your actual port (e.g., '/dev/ttyUSB0' on Linux/Mac)
SERIAL_PORT = 'COM8' 
BAUD_RATE = 115200

def send_command(ser, command_char, value):
    """Sends a command like 'I50' to the STM32"""
    # Create string like "I50\n"
    command_str = f"{command_char}{value}\n"
    
    # Send it (encoded to bytes)
    ser.write(command_str.encode())
    print(f"Sent: {command_str.strip()}")
    
    # Wait a tiny bit for STM32 to reply
    time.sleep(0.1)
    
    # Read the reply from STM32
    while ser.in_waiting:
        response = ser.readline().decode().strip()
        print(f"STM32 Replied: {response}")
```

```python
ser = serial.Serial(SERIAL_PORT, BAUD_RATE, timeout=1)
```

```python
send_command(ser, 'T',150)
send_command(ser, 'K',80)
send_command(ser, 'I',750)
```
