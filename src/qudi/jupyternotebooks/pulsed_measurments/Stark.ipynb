```python
import importlib, keithley2600
print(keithley2600.__file__)   # confirm which file is loaded
importlib.reload(keithley2600)
smu = keithley2600.SMU26xx("USB0::0x05E6::0x2636::4365643::INSTR")
smu_a = smu.get_channel('a')

import time
from pyvisa.errors import VisaIOError
import numpy as np
from matplotlib import pyplot as plt
import time
from toptica.lasersdk.dlcpro.v2_0_3 import DLCpro, NetworkConnection

def set_voltage(v, retries=6, delay_s=0.2):
    last_exc = None
    for _ in range(retries):
        try:
            smu_a.set_voltage(float(v))
            return True
        except VisaIOError as e:
            last_exc = e
            time.sleep(delay_s)
    raise RuntimeError(f"set_voltage failed at {v} V: {last_exc}")

def get_current(current_v=None, retries=3, delay_s=0.2, reassert_voltage=False):
    last_exc = None
    for _ in range(retries):
        try:
            return float(smu_a.measure_current())
        except VisaIOError as e:
            last_exc = e
            if reassert_voltage and current_v is not None:
                set_voltage(current_v, retries=1, delay_s=delay_s)
            time.sleep(delay_s)
    raise RuntimeError(f"get_current failed at {current_v} V: {last_exc}")
```
