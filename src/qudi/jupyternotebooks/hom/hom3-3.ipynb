```python
# pulsestreamer._seq.setDigital(1, [(1000, 1)])
T_pulse = 100 * 1e6 #mu sec 
T_wait = 1 * 1e6
T_repump = 10*1e6
T_total = T_pulse + T_wait + T_repump + T_wait
for sp in [-1.37, -0.8, -0.4, 0, .4, 0.8, 1.37]:
    # start measurement trigger
    hom.pulsestreamer._seq.setDigital(5, [(T_total, 0), (10, int(True))])
    # start resonant laser pulse
    hom.pulsestreamer._seq.setDigital(1, [(T_pulse, int(True)), (T_total - T_pulse, 0)])
    # repump
    hom.pulsestreamer._seq.setDigital(3, [(T_pulse + T_wait, int(False)), (T_repump, int(True)), (T_wait, int(False))])

    hom.pulsestreamer.pulser_on()
#trigger the histogram
```
