```python
def refocus():
    global iteration, OFFSET, do_realign
    iteration += 1
    if iteration % 2:
        hom.measurement_mode('Off-res')
        hom.polarization_is_parallel = False
        hom.refocus(remote=True, local=True)
        time.sleep(0.2)
        hom.polarization_is_parallel = True
        
        
    if iteration % 1:
        if do_realign:
            tagger.tagger.setConditionalFilter(trigger=[5], filtered=[2,3])
            hom.align_resonances(dv=0.3, align_iter = 3, offset=OFFSET)
            time.sleep(3)
            hom.measurement_mode('Off-res')

            time.sleep(1)

            hom.polarization_is_parallel = True

            tagger.tagger.setConditionalFilter(trigger=[5], filtered=[1])
    
    hom.measurement_mode('Off-res')
    time.sleep(0.2)
    hom.polarization_is_parallel = True
```
