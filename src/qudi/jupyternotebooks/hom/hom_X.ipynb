```python
import ast
def get_file_data_n_info(file_path):
    _data = np.genfromtxt(os.path.join('', file_path))

    # First, let's re-open the file to extract the header information related to the axis ticks.
    header_info = {}

    with open(file_path, 'r') as file:
        for line in file:
            # Looking for axis name and unit information in the header
            if line.startswith("# a axis name"):
                header_info['a_axis_name'] = line.split('=')[1].strip()
            elif line.startswith("# a axis unit"):
                header_info['a_axis_unit'] = line.split('=')[1].strip()
            elif line.startswith("# a scan range"):
                header_info['a_scan_range'] = ast.literal_eval(line.split('=')[1].strip())
            elif line.startswith("# a axis resolution"):
                header_info['a_resolution'] = int(line.split('=')[1].strip())

    # Displaying extracted header information for verification
    return _data, header_info
```
