```python
def wait_for_B():
    time.sleep(1)
    while min(vector_magnet.get_ramping_state()) == 1:
        time.sleep(1)
```

```python
def repump(Laser='Laser_520', pulse_length=0.5):
    switchlogic.set_state(Laser, 'on')
    time.sleep(pulse_length)
    switchlogic.set_state(Laser, 'off')
```

```python
def save_ple(tag, poi_name=None, folder_name = None):
        if folder_name:
            ple_gui._save_folderpath = folder_name
        ple_gui.save_path_widget.saveTagLineEdit.setText(
            f"{poi_name}_{tag}"
            )
        ple_gui._mw.actionSave.triggered.emit()

def refocus_all():
    ple_optimize_logic.toggle_optimize(True)
    while ple_optimize_logic.optimizer_running:
            time.sleep(1)
    repump()
    switchlogic.set_state("Laser_620", 'on')
    scanning_optimize_logic.toggle_optimize(True)
    while scanning_optimize_logic.optimizer_running:
            time.sleep(1)
    switchlogic.set_state("Laser_620", 'off')
```

```python
Measurments = len(phi_sweep)*len(theta_sweep)*len(B_sweep)
i = 0
magnetlogic.ramp([0.2, 0, 0])
wait_for_B()
for i_phi, phi in enumerate(phi_sweep):
    for i_theta, theta in enumerate(theta_sweep):
        for i_B, B in enumerate(B_sweep):
            magnetlogic.ramp([B, theta, phi])
            wait_for_B()
            ple_scanner_logic.update_number_of_repeats(5)
            refocus_all()
            ple_scanner_logic.update_number_of_repeats(10)
            ple_scanner_logic.start_scan([ple_scanner_logic._scan_axis])
            while ple_scanner_logic.module_state() != 'idle':
                time.sleep(1)
                # repump()
            save_ple(f'phi={int(phi)}_theta={int(theta)}_B={B}')
            #curr_data = laser_scanner_logic.scan_data.to_dict()['data']['APD1']
            #final_data[i_theta, i_phi] = curr_data
            i = i +1
            print(f"{i} from {Measurments} done")
magnetlogic.ramp_to_zero()
```

```python
# Ramp and refocus
laser_scanner_logic.update_number_of_repeats(50)
B_sweep = np.linspace(0.4, 0, 20)

for i_B, B in enumerate(B_sweep):
    magnetlogic.ramp([B, 80, 0])
    wait_for_B()
    laser_scanner_logic.start_scan([laser_scanner_logic._scan_axis])
    while laser_scanner_logic.module_state() != 'idle':
        time.sleep(1)
    save_ple(f'B={B}_theta=80')
    # process_setpoint_combiner.set_setpoint('LaserScanner_red', 0)
    # time.sleep(2)
    # scanning_optimize_logic.start_optimize()
    while scanning_optimize_logic.module_state() != 'idle':
            time.sleep(1)

```
